```python
# Carregando modelo de análise de sentimento
classificador = pipeline("sentiment-analysis")

# Lista de frases para teste
frases = [
    ("O dia está maravilhoso e tudo parece perfeito.", "Positiva"),
    ("Estou profundamente frustrado com esse serviço horrível.", "Negativa"),
    ("O policial viu o homem com a arma.", "Ambígua"),
    ("A reunião de hoje foi tão produtiva que poderia ter sido um e-mail.", "Irônica"),
    ("A síntese proteica ocorre nos ribossomos da célula.", "Outro domínio")
]

resultados_classificacao = []
for frase, tipo in frases:
    res = classificador(frase)[0]
    resultados_classificacao.append({
        "Tipo": tipo,
        "Frase": frase,
        "Classe": res['label'],
        "Score": round(res['score'], 4),
        "Análise": "..." # Preencher após rodar
    })

df_class = pd.DataFrame(resultados_classificacao)
df_class
```

```python
# Carregando modelo de geração
gerador = pipeline("text-generation", model="distilgpt2")

# Testes
# 1. Prompt curto
res1 = gerador("Artificial intelligence can", max_new_tokens=30)
# 2. Prompt detalhado
res2 = gerador("Explain the benefits of renewable energy for a sustainable future in one paragraph.", max_new_tokens=50)
# 3. Temperatura Baixa (0.2)
res3 = gerador("The future of technology is", max_new_tokens=30, do_sample=True, temperature=0.2)
# 4. Temperatura Alta (1.2)
res4 = gerador("The future of technology is", max_new_tokens=30, do_sample=True, temperature=1.2)
# 5. Duas sequências
res5 = gerador("Learning to code is", max_new_tokens=30, do_sample=True, temperature=0.8, num_return_sequences=2)

# Exiba os resultados para copiar para o seu PDF
print(f"Curto: {res1[0]['generated_text']}")
print(f"Detalhado: {res2[0]['generated_text']}")
print(f"Temp 0.2: {res3[0]['generated_text']}")
print(f"Temp 1.2: {res4[0]['generated_text']}")
print(f"Duas Seq: {res5}")
```
